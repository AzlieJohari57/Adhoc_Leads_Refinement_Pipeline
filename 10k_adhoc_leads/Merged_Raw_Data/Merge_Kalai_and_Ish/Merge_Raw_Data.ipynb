### Merge Adhoc Leads _F&B.xlsx Into One Tab

```python
# Path to your Excel file
file_path = "Adhoc Leads _F&B.xlsx"

# --- Function to clean & format SG numbers (8 digits only) ---
def clean_sg_number(val):
    if pd.isna(val):
        return None
    
    # Extract only digits
    digits = re.sub(r"\D", "", str(val))
    
    # Remove leading "65" if it's a country code
    if digits.startswith("65") and len(digits) > 8:
        digits = digits[2:]
    
    # Valid Singapore mobile/landline numbers start with 6, 8, or 9 and are 8 digits
    if len(digits) == 8 and digits[0] in ["6", "8", "9"]:
        return digits
    
    return None


# Load the Excel workbook
excel = pd.ExcelFile(file_path)

dfs = []

# Loop through each sheet
for sheet in excel.sheet_names:
    df = excel.parse(sheet)

    # Standardize column names
    df.columns = df.columns.str.strip().str.lower()

    # Reformat Singapore contact numbers
    for i in range(1, 9):
        col = f"contact_number {i}"
        if col in df.columns:
            df[col] = df[col].apply(clean_sg_number)

    dfs.append(df)

# Merge all sheets
merged_df = pd.concat(dfs, ignore_index=True, sort=False)

# Select only the desired columns
selected_columns = [
    "company_name",
    "pic_name",
    "designation",
    "contact_number 1",
    "contact_number 2",
    "contact_number 3",
    "contact_number 4",
    "contact_number 5",
    "contact_number 6",
    "contact_number 7",
    "contact_number 8",
    "email_address",
    "address"
]

# Keep only existing columns
merged_df = merged_df[[col for col in selected_columns if col in merged_df.columns]]

# Save to new Excel file
merged_df.to_excel("Merged_Adhoc_Leads_F&B.xlsx", index=False)

print("✅ Merged file created with cleaned 8-digit SG numbers (no country code).")
```

### Check for Duplication with Master DB

```python
# -------------------------------
# 1. Helper function: clean SG number
# -------------------------------
def clean_sg_number(val):
    if pd.isna(val):
        return None

    # Convert float like 65131787.0 to 65131787
    if isinstance(val, float):
        val = str(int(val))
    else:
        val = str(val)

    # keep digits only
    digits = re.sub(r"\D", "", val)

    # remove leading 65 (country code)
    if digits.startswith("65") and len(digits) > 8:
        digits = digits[2:]

    # valid SG number = 8 digits, starts with 6/8/9
    if len(digits) == 8 and digits[0] in ["6", "8", "9"]:
        return digits

    return None

```

```python

# -------------------------------
# 2. File 1: Merged_Multiple_Raw_Data_Files.xlsx
# -------------------------------

file1 = "Merged_Multiple_Raw_Data_Files.xlsx"

file1_cols = [
    "contact_number 1", "contact_number 2", "contact_number 3", "contact_number 4",
    "contact_number 5", "contact_number 6", "contact_number 7", "contact_number 8"
]

print("STEP A: Loaded file")


df1 = pd.read_excel(file1)

# Standardize column names
df1.columns = df1.columns.str.strip().str.lower()

print("STEP B: Columns standardized")


print(df1.columns.tolist())

# Extract only columns that exist
file1_existing = [c for c in file1_cols if c in df1.columns]

print("STEP C: Columns found:", file1_existing)


print("STEP D: Showing raw numbers")
print(df1[file1_existing].head(20))

print("STEP E: Start cleaning")

# Flatten into a Series of cleaned numbers
numbers_file1 = (
    df1[file1_existing]
    .apply(lambda col: col.map(clean_sg_number))
    .values
    .ravel()
)

numbers_file1 = pd.Series([n for n in numbers_file1 if n is not None], name="number")
print(numbers_file1.head(20))
```

```python

# -------------------------------
# 3. File 2: [18 Nov] Master DB.xlsx
# -------------------------------

file2 = "[18 Nov] Master DB.xlsx"

file2_cols = [
    "pic name 1 contact number 1", "pic name 1 contact number 2", 
    "pic name 1 contact number 3", "pic name 1 contact number 4",
    
    "pic name 2 contact number 1", "pic name 2 contact number 2",
    "pic name 2 contact number 3", "pic name 2 contact number 4",

    "pic name 3 contact number 1", "pic name 3 contact number 2",
    "pic name 3 contact number 3", "pic name 3 contact number 4"
]

df2 = pd.read_excel(file2)

# Standardize column names
df2.columns = df2.columns.str.strip().str.lower()

# Extract existing columns
file2_existing = [c for c in file2_cols if c in df2.columns]

# Flatten into a Series of cleaned numbers
numbers_file2 = (
    df2[file2_existing]
    .applymap(clean_sg_number)
    .values
    .ravel()
)

numbers_file2 = pd.Series([n for n in numbers_file2 if n is not None], name="number")
```

```python

# -------------------------------
# 4. Find duplicates across the two files
# -------------------------------

# Convert to sets to identify duplicates
duplicates = set(numbers_file1).intersection(set(numbers_file2))

# Prepare result dataframe
result_df = pd.DataFrame({
    "number": list(duplicates),
    "count_in_file1": [sum(numbers_file1 == num) for num in duplicates],
    "count_in_file2": [sum(numbers_file2 == num) for num in duplicates],
})

# Sort result
result_df = result_df.sort_values("number").reset_index(drop=True)
```

```python

# -------------------------------
# 5. Export to Excel
# -------------------------------

result_df.to_excel("Duplicated_Contact_Numbers.xlsx", index=False)

print("✅ Duplicate contact-number check completed!")
print(f"Total duplicated numbers found: {len(result_df)}")
print("📁 Output saved to: Duplicated_Contact_Numbers.xlsx")
```

### Apply Constraints to Show Only Relevant Leads (SG only, no competitors, no delivery platforms, cleaned company name)

1. Email address: 
- Keeps only rows with email address ending with "sg" or "com"
- Remove rows where the “company_name” column is blank and the “email_address” column contains only personal email addresses (with the “google” or “yahoo” or “singnet” domain)
- Remove rows containing these terms in the "email_address" column: “marketing”, “entertainment” 
- Remove rows containing "deliveryhero", "tabsquare" or "foodpanda" in the "email_address" column

2. Company name: 
- Remove these texts from the “company_name” column: “SmartTerminals”, “deal”, “- Raptor List (Pay Only)”, “Incoming Lead_”, “FHA 2022_”, “EDM leads _”, “Google Ads Leads_”, “Speedster Campaign_”, “Fy23_”, “TabsquareConneXions_”, “QSR 2024 -”, “Pre Event 2024-”, “(SmartConnect)”, “FP23_”, “FY23_SG_RA_”, “TabsquareConneXions_”, “.. “
- Remove these symbols (“_”, “*”, “|”) from the “company_name” column values
- Remove rows where the "company_name" column contain "marketing" or "consulting"

3. Competitor: Exclude a row if the email address contains POS vendors' name
- suntoyo
- megasafe    
- appzgate
- Qashier
- Weebo
- Azdigital
- Hashmicro
- Netiquette
- Edgeworks
- Foodzaps
- Sgdatahub
- Imetrics
possquare

4. Remove rows containing duplicated contact numbers (within the output) 

```python
import pandas as pd

# -------------------------------
# Load the Excel file
# -------------------------------
file_path = "Merged_Multiple_Raw_Data_Files.xlsx"
df = pd.read_excel(file_path)

# Standardize column names
df.columns = df.columns.str.strip().str.lower()

# -------------------------------
# 1️⃣ Email Address Constraints
# -------------------------------

# Keep only emails ending with .sg or .com
df = df[df['email_address'].str.lower().str.endswith(('.sg', '.com'), na=False)]

# Remove rows where company_name is blank AND personal email
personal_domains = ['google', 'yahoo', 'singnet']
mask_personal_email = df['company_name'].isna() & df['email_address'].str.lower().str.contains('|'.join(personal_domains), na=False)
df = df[~mask_personal_email]

# Remove rows with marketing or entertainment in email
df = df[~df['email_address'].str.lower().str.contains('marketing|entertainment', na=False)]

# Remove rows with delivery platforms
delivery_platforms = ['deliveryhero', 'tabsquare', 'foodpanda']
df = df[~df['email_address'].str.lower().str.contains('|'.join(delivery_platforms), na=False)]

# -------------------------------
# 2️⃣ Company Name Constraints
# -------------------------------

# Remove unwanted texts from company_name
unwanted_texts = [
    "SmartTerminals", "deal", "- Raptor List (Pay Only)", "Incoming Lead_", "FHA 2022_",
    "EDM leads _", "Google Ads Leads_", "Speedster Campaign_", "Fy23_", "TabsquareConneXions_",
    "QSR 2024 -", "Pre Event 2024-", "(SmartConnect)", "FP23_", "FY23_SG_RA_", "TabsquareConneXions_", ".. "
]
for text in unwanted_texts:
    df['company_name'] = df['company_name'].str.replace(text, '', case=False, regex=False)

# Remove unwanted symbols
df['company_name'] = df['company_name'].str.replace(r'[_*|]', '', regex=True)

# Remove rows containing "marketing" or "consulting" in company_name
df = df[~df['company_name'].str.lower().str.contains('marketing|consulting', na=False)]

# -------------------------------
# 3️⃣ Competitor Exclusion
# -------------------------------
competitors = [
    "suntoyo", "megasafe", "appzgate", "qashier", "weebo", "azdigital",
    "hashmicro", "netiquette", "edgeworks", "foodzaps", "sgdatahub", "imetrics", "possquare"
]

df = df[~df['email_address'].str.lower().str.contains('|'.join(competitors), na=False)]

# -------------------------------
# 4️⃣ Remove Duplicate Contact Numbers (keep first occurrence)
# -------------------------------

contact_cols = [f"contact_number {i}" for i in range(1, 9) if f"contact_number {i}" in df.columns]

seen_numbers = set()
rows_to_keep = []

for idx, row in df.iterrows():
    row_numbers = row[contact_cols].dropna().astype(str).tolist()
    if any(num in seen_numbers for num in row_numbers):
        continue  # skip row if any number is already seen
    else:
        seen_numbers.update(row_numbers)
        rows_to_keep.append(idx)

df = df.loc[rows_to_keep]

# -------------------------------
# 5️⃣ Keep only desired columns
# -------------------------------
columns_to_keep = [
    "company_name",
    "pic_name",
    "designation",
    "contact_number 1",
    "contact_number 2",
    "contact_number 3",
    "contact_number 4",
    "contact_number 5",
    "contact_number 6",
    "contact_number 7",
    "contact_number 8",
    "email_address",
    "address"
]

# Filter only the columns that exist in the dataframe
columns_to_keep_existing = [col for col in columns_to_keep if col in df.columns]
df = df[columns_to_keep_existing]

# -------------------------------
# Save the cleaned data
# -------------------------------
output_file = "Merged_Multiple_Raw_Data_Files_Cleaned.xlsx"
df.to_excel(output_file, index=False)

print(f"Cleaning complete! Saved to {output_file}")
```

### Reformatting contact number to Master DB format (65 XXXX XXXX)

```python
# -------------------------------
# Load the merged Excel file
# -------------------------------
df = pd.read_excel("Merged_Multiple_Raw_Data_Files_Cleaned.xlsx")

# Standardize column names
df.columns = df.columns.str.strip().str.lower()

# -------------------------------
# Standardize SG numbers to 65 XXXX XXXX format
# -------------------------------
def format_sg_number(number):
    number = re.sub(r'\D', '', str(number))  # remove non-digits
    if number.startswith('65') and len(number) == 10:
        number = number[2:]  # remove '65' prefix
    elif len(number) == 8:
        pass  # already local 8 digits
    else:
        return None  # invalid number
    return f"65 {number[:4]} {number[4:]}"

# List of contact number columns
contact_cols = [f"contact_number {i}" for i in range(1, 9) if f"contact_number {i}" in df.columns]

# Apply formatting to each contact number column
for col in contact_cols:
    df[col] = df[col].apply(format_sg_number)

# -------------------------------
# Save the standardized data
# -------------------------------
output_file = "Merged_Multiple_Raw_Data_Files_Cleaned_Reformatted.xlsx"
df.to_excel(output_file, index=False)

print(f"✅ Contact numbers standardized in columns {contact_cols}. Saved as '{output_file}'")
```
