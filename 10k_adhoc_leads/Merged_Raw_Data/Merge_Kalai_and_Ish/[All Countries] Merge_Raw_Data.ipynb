```python
# -------------------------------
# Load the Excel file
# -------------------------------
file_path = r"C:\Users\teele\Documents\GitHub\Adhoc_Leads_Refinement_Pipeline\10k_adhoc_leads\Merged_Raw_Data\Merge_Kalai_and_Ish\Merged_Multiple_Raw_Data_Files_NoDuplicates.xlsx"
df = pd.read_excel(file_path)

# Standardize column names
df.columns = df.columns.str.strip().str.lower()

# -------------------------------
# 1️⃣ Email Address Constraints
# -------------------------------

# # Keep only emails ending with .sg or .com
# df = df[df['email_address'].str.lower().str.endswith(('.sg', '.com'), na=False)]

# # Remove rows where company_name is blank AND personal email
# personal_domains = ['gmail', 'yahoo', 'ymail', 'singnet', 'hotmail', 'outlook']
# mask_personal_email = df['company_name'].isna() & df['email_address'].str.lower().str.contains('|'.join(personal_domains), na=False)
# df = df[~mask_personal_email]

# Remove rows with marketing or entertainment in email address
df = df[~df['email_address'].str.lower().str.contains('marketing|entertainment', na=False)]

# Remove rows with delivery platforms in email address
delivery_platforms = ['deliveryhero', 'tabsquare', 'foodpanda', "grab"]
df = df[~df['email_address'].str.lower().str.contains('|'.join(delivery_platforms), na=False)]

# -------------------------------
# 2️⃣ Company Name Constraints
# -------------------------------

# Remove unwanted texts from company_name
unwanted_texts = [
    "SmartTerminals", "deal", "- Raptor List (Pay Only)", "Incoming Lead_", "FHA 2022_",
    "EDM leads _", "Google Ads Leads_", "Speedster Campaign_", "Fy23_", "TabsquareConneXions_",
    "QSR 2024 -", "Pre Event 2024-", "(SmartConnect)", "FP23_", "FY23_SG_RA_", "TabsquareConneXions_", ".. "
]
for text in unwanted_texts:
    df['company_name'] = df['company_name'].str.replace(text, '', case=False, regex=False)

# Remove unwanted symbols
df['company_name'] = df['company_name'].str.replace(r'[_*|]', '', regex=True)

# Remove rows containing "marketing" or "consulting" in company_name
df = df[~df['company_name'].str.lower().str.contains('marketing|consulting', na=False)]

# -------------------------------
# 3️⃣ Competitor Exclusion (from the email address)
# -------------------------------
competitors = [
    "suntoyo", "megasafe", "appzgate", "qashier", "weebo", "azdigital",
    "hashmicro", "netiquette", "edgeworks", "foodzaps", "sgdatahub", "imetrics", "possquare"
]

df = df[~df['email_address'].str.lower().str.contains('|'.join(competitors), na=False)]

# -------------------------------
# 4️⃣ Remove Duplicate Contact Numbers (keep first occurrence)
# -------------------------------

contact_cols = [f"contact_number {i}" for i in range(1, 9) if f"contact_number {i}" in df.columns]

seen_numbers = set()
rows_to_keep = []

for idx, row in df.iterrows():
    row_numbers = row[contact_cols].dropna().astype(str).tolist()
    if any(num in seen_numbers for num in row_numbers):
        continue  # skip row if any number is already seen
    else:
        seen_numbers.update(row_numbers)
        rows_to_keep.append(idx)

df = df.loc[rows_to_keep]

# -------------------------------
# 5️⃣ Remove duplicate email addresses for blank company_name
# -------------------------------
# Keep first occurrence only
mask_blank_company = df['company_name'].isna()
df_blank_company = df[mask_blank_company].drop_duplicates(subset=['email_address'], keep='first')
df_non_blank_company = df[~mask_blank_company]
df = pd.concat([df_non_blank_company, df_blank_company], ignore_index=True)

# -------------------------------
# 6️⃣ Split "email_address" into multiple columns
# -------------------------------
# Assuming emails are separated by commas
email_split = df['email_address'].str.split(',', expand=True)
email_split.columns = [f'email_{i+1}' for i in range(email_split.shape[1])]

# Concatenate split email columns to the original dataframe and drop original column
df = pd.concat([df.drop(columns=['email_address']), email_split], axis=1)

# -------------------------------
# 7️⃣ Format text columns
# -------------------------------
# Convert company_name to uppercase
df['company_name'] = df['company_name'].str.upper()

# Capitalize PIC names (first letter of each word)
df['pic_name'] = df['pic_name'].str.title()  # str.title() capitalizes first letter of each word

# -------------------------------
# 8️⃣ Keep only desired columns
# -------------------------------
columns_to_keep = [
    "company_name",
    "pic_name",
    "designation",
    "contact_number 1",
    "contact_number 2",
    "contact_number 3",
    "contact_number 4",
    "contact_number 5",
    "contact_number 6",
    "contact_number 7",
    "email_address",
    "address"
]

# Include all split email columns
email_columns = [col for col in df.columns if col.startswith('email_')]
columns_to_keep += email_columns

columns_to_keep_existing = [col for col in columns_to_keep if col in df.columns]
df = df[columns_to_keep_existing]

# -------------------------------
# Save the cleaned data
# -------------------------------
output_file = "Merged_Multiple_Raw_Data_Files_NoDuplicates_Cleaned.xlsx"
df.to_excel(output_file, index=False)

print(f"Cleaning complete! Saved to {output_file}")
```

### Categorise contact number according to their country of origin

```python
# --- Load your Excel file ---
file_path = r"C:\Users\teele\Documents\GitHub\Adhoc_Leads_Refinement_Pipeline\10k_adhoc_leads\Merged_Raw_Data\Merge_Kalai_and_Ish\Merged_Multiple_Raw_Data_Files_NoDuplicates_Cleaned.xlsx"
sheet_name = "Sheet1"
df = pd.read_excel(file_path, sheet_name=sheet_name)

# List of contact number columns
contact_cols = [
    "contact_number 1",
    "contact_number 2",
    "contact_number 3",
    "contact_number 4",
    "contact_number 5",
    "contact_number 6",
    "contact_number 7"
]

# --- Step 1: Clean numbers ---
def clean_number(number):
    if pd.isna(number):
        return ""
    return re.sub(r"\D", "", str(number))

# --- Step 2: Country dial codes ---
country_dial_codes = {
    "Singapore": "65",
    "Malaysia": "60",
    "Japan": "81",
    "Australia": "61",
    "China": "86",
    "Poland": "48",
    "Canada / USA": "1",
    "United Kingdom": "44",
    "India": "91",
    "Vietnam": "84",
    "Italy": "39",
    "Mongolia": "976",
    "British Indian Ocean Territory": "246",
    "Montenegro": "382",
    "Hong Kong": "852",
    "Belgium": "32",
    "South Africa": "27",
    "Libya": "218",
    "Ireland": "353",
    "Netherlands": "31",
    "Saudi Arabia": "966",
    "Taiwan": "886",
    "Anguilla": "264",
    "Indonesia": "62",
    "Saint Vincent & Grenadines": "784",
    "Philippines": "63",
    "Mexico": "52",
    "Thailand": "66",
    "Cambodia": "855",
    "Brunei": "673"
}

# --- Step 3: Detect country ---
def detect_country(number):
    if number.startswith("65") or (len(number) == 8 and number[0] in ["6","8","9"]):
        return "Singapore"
    elif number.startswith("60") or (len(number) in [9,10] and number.startswith("1")):
        return "Malaysia"
    elif number.startswith("81") or (len(number) == 10 and number.startswith("0")):
        return "Japan"
    elif number.startswith("61") or (len(number) == 9 and number.startswith("4")):
        return "Australia"
    elif number.startswith("86") or (len(number) == 11 and number.startswith("1")):
        return "China"
    elif len(number) == 9 and number[0] in ["5","6","7","8","9"]:
        return "Poland"
    elif len(number) == 10:
        return "Canada / USA"
    elif number.startswith("44"):
        return "United Kingdom"
    elif number.startswith("91"):
        return "India"
    elif number.startswith("84"):
        return "Vietnam"
    elif number.startswith("39"):
        return "Italy"
    elif number.startswith("976"):
        return "Mongolia"
    elif number.startswith("246"):
        return "British Indian Ocean Territory"
    elif number.startswith("382"):
        return "Montenegro"
    elif number.startswith("852"):
        return "Hong Kong"
    elif number.startswith("32"):
        return "Belgium"
    elif number.startswith("27"):
        return "South Africa"
    elif number.startswith("218"):
        return "Libya"
    elif number.startswith("353"):
        return "Ireland"
    elif number.startswith("31"):
        return "Netherlands"
    elif number.startswith("966"):
        return "Saudi Arabia"
    elif number.startswith("886"):
        return "Taiwan"
    elif number.startswith("264"):
        return "Anguilla"
    elif number.startswith("62"):
        return "Indonesia"
    elif number.startswith("784"):
        return "Saint Vincent & Grenadines"
    elif number.startswith("63"):
        return "Philippines"
    elif number.startswith("52"):
        return "Mexico"
    elif number.startswith("66"):
        return "Thailand"
    elif number.startswith("855"):
        return "Cambodia"
    elif number.startswith("673"):
        return "Brunei"
    else:
        return "Unknown"

# --- Step 4: Reformat numbers ---
def reformat_number(number, country):
    if not number:
        return ""
    
    number_only = number.lstrip("0")
    
    if country == "Singapore":
        if number_only.startswith("65"):
            number_only = number_only[2:]
        if len(number_only) == 8:
            return f"65 {number_only[:4]} {number_only[4:]}"
        else:
            return f"65 {number_only}"
    else:
        dial_code = country_dial_codes.get(country, "")
        if number_only.startswith(dial_code):
            number_only = number_only[len(dial_code):]
        return f"{dial_code} {number_only}"

# --- Step 5: Apply to all contact columns ---
for col in contact_cols:
    cleaned_col = f"{col}_clean"
    formatted_col = f"{col}_formatted"
    country_col = f"{col}_country"
    
    # Clean
    df[cleaned_col] = df[col].apply(clean_number)
    
    # Detect country
    df[country_col] = df[cleaned_col].apply(detect_country)
    
    # Reformat
    df[formatted_col] = df.apply(lambda row: reformat_number(row[cleaned_col], row[country_col]), axis=1)

# --- Step 6: Save results ---
output_file = "Merged_Multiple_Raw_Data_Files_NoDuplicates_Cleaned_Contact Number Reformatted.xlsx"
df.to_excel(output_file, index=False)
print(f"✅ Saved categorized and formatted contacts for all columns to {output_file}")
```
