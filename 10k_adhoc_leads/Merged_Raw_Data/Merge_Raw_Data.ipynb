### Merge Adhoc Leads _F&B.xlsx Into One Tab

```python
# Path to your Excel file
file_path = "Adhoc Leads _F&B.xlsx"

# Load the Excel workbook
excel = pd.ExcelFile(file_path)

# List to store all sheets' data
dfs = []

# Loop through each sheet
for sheet in excel.sheet_names:
    df = excel.parse(sheet)

    # --- Standardize column names ---
    df.columns = df.columns.str.strip().str.lower()

    dfs.append(df)

# --- Merge all sheets ---
merged_df = pd.concat(dfs, ignore_index=True, sort=False)

# --- Select only the desired columns ---
selected_columns = [
    "company_name",
    "pic_name",
    "designation",
    "contact_number",
    "email_address",
    "address"
]

# Keep only the columns that exist in the merged file
merged_df = merged_df[[col for col in selected_columns if col in merged_df.columns]]

# --- Save to new Excel file ---
merged_df.to_excel("Merged_Adhoc_Leads_F&B.xlsx", index=False)

print("✅ Merged file created: Merged_Adhoc_Leads_F&B.xlsx")
```

### Apply Constraints to Show Only Relevant Leads (SG only, no competitors, no delivery platforms)

1. Contact number: keeps only rows with SG contact number 
2. Email address: keeps only rows with email address ending with "sg" or "com"
3. Competitor: Exclude a row if the email address contains POS vendors' name
- suntoyo
- megasafe    
- appzgate
- Qashier
- Weebo
- Azdigital
- Hashmicro
- Netiquette
- Edgeworks
- Foodzaps
- Sgdatahub
- Imetrics
possquare
4. Exclude a row if the email address contains "deliveryhero", "tabsquare" or "foodpanda"
5. Exclude a row if the company_name contains "marketing" or "consulting"
6. Drop duplicates based on contact_number

```python
import pandas as pd
import re

# Load the merged Excel file
df = pd.read_excel("Merged_Multiple_Raw_Data_Files.xlsx")

# Standardize column names
df.columns = df.columns.str.strip().str.lower()

# 1️⃣ Singapore phone numbers (allow spaces, dashes, parentheses)
df['contact_number_clean'] = df['contact_number'].astype(str).str.replace(r'\D', '', regex=True)

# Keep only numbers with 8 digits (local) or 65+8 digits (with country code)
sg_phone_pattern = r'^(?:65)?\d{8}$'
df = df[df['contact_number_clean'].str.match(sg_phone_pattern, na=False)]

# 1️⃣a Standardize to 65 XXXX XXXX format
def format_sg_number(number):
    number = re.sub(r'\D', '', str(number))  # remove non-digits
    if number.startswith('65') and len(number) == 10:
        number = number[2:]  # remove '65' prefix
    elif len(number) == 8:
        pass  # already local 8 digits
    else:
        return None  # shouldn't happen due to filtering
    return f"65 {number[:4]} {number[4:]}"

df['contact_number'] = df['contact_number_clean'].apply(format_sg_number)
df.drop(columns=['contact_number_clean'], inplace=True)

# 2️⃣ Prepare lowercased emails (NaN-safe)
df['email_lower'] = df['email_address'].fillna("").str.lower()

# 2️⃣ Keep only emails ending with 'sg' or 'com'
df = df[df['email_lower'].str.endswith("sg") | df['email_lower'].str.endswith("com")]

# 3️⃣ Remove rows with excluded keywords in email
exclude_keywords = [
    "suntoyo", "megasafe", "appzgate", "qashier", "weebo", "azdigital",
    "hashmicro", "netiquette", "edgeworks", "foodzaps", "sgdatahub",
    "imetrics", "possquare", "deliveryhero", "tabsquare", "foodpanda"
]
pattern = '|'.join(exclude_keywords)
df = df[~df['email_lower'].str.contains(pattern, na=False)]

# Drop helper column
df.drop(columns=['email_lower'], inplace=True)

# 4️⃣ Exclude rows where company_name contains "marketing" or "consulting"
df = df[~df['company_name'].str.contains(r"marketing|consulting", case=False, na=False)]

# 5️⃣ Remove duplicates based on contact_number
df = df.drop_duplicates(subset=['contact_number'], keep='first')

# Save filtered, standardized, and deduplicated data
df.to_excel("Filtered_Standardized_Deduped_Raw_Data.xlsx", index=False)

print("✅ Filtering, standardization, and deduplication complete — excluding rows with 'marketing' or 'consulting' in company_name. Saved as 'Filtered_Standardized_Deduped_Raw_Data.xlsx'")
```
